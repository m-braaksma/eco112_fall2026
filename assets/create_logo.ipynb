```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle

# ------------------------------------------------------------------
# Augsburg colors
# ------------------------------------------------------------------

MAROON = "#660033"
MAROON_LIGHT = "#74233d"
SILVER = "#a9a9aa"
LIGHT_GRAY = "#dddddd"

# ------------------------------------------------------------------
# Figure
# ------------------------------------------------------------------

fig, ax = plt.subplots(figsize=(6, 6), facecolor="white")

ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.set_aspect("equal")

# ------------------------------------------------------------------
# Globe
# ------------------------------------------------------------------

center_y = 4.8
center_x = 5
radius = 3.0

# Outer globe
globe = Circle(
    (center_x, center_y),
    radius,
    fill=False,
    linewidth=2.5,
    edgecolor=MAROON,
    alpha=0.9,
)

ax.add_patch(globe)

# ------------------------------------------------------------------
# Latitude lines
# ------------------------------------------------------------------

theta = np.linspace(0, 2 * np.pi, 300)

for scale in [0.45, 0.75]:

    x = center_x + radius * np.cos(theta)
    y = center_y + radius * scale * np.sin(theta)

    ax.plot(
        x,
        y,
        color=SILVER,
        linewidth=1.2,
        alpha=0.8,
    )

# ------------------------------------------------------------------
# Longitude lines
# ------------------------------------------------------------------

for scale in [0.45, 0.75]:

    x = center_x + radius * scale * np.cos(theta)
    y = center_y + radius * np.sin(theta)

    ax.plot(
        x,
        y,
        color=SILVER,
        linewidth=1.2,
        alpha=0.8,
    )

# Equator
ax.plot(
    [center_x - radius, center_x + radius],
    [center_y, center_y],
    color=SILVER,
    linewidth=1.2,
    alpha=0.8,
)

# Vertical center line
ax.plot(
    [center_x, center_x],
    [center_y - radius, center_y + radius],
    color=SILVER,
    linewidth=1.2,
    alpha=0.8,
)

# ------------------------------------------------------------------
# Economic growth line across the globe
# ------------------------------------------------------------------

x = np.linspace(2.4, 7.6, 100)

y = (
    center_y
    - 0.9
    + 0.28 * (x - 2.4)
    + 0.18 * np.sin((x - 2.4) * 2.2)
)

# Clip approximately to globe boundaries visually
ax.plot(
    x,
    y,
    color=MAROON,
    linewidth=3,
    alpha=0.95,
)

# Small data points along the trend
x_pts = np.linspace(2.8, 7.2, 7)

y_pts = (
    center_y
    - 0.9
    + 0.28 * (x_pts - 2.4)
    + 0.18 * np.sin((x_pts - 2.4) * 2.2)
)

ax.scatter(
    x_pts,
    y_pts,
    color=MAROON_LIGHT,
    s=35,
    alpha=0.9,
    zorder=3,
)

# ------------------------------------------------------------------
# Text
# ------------------------------------------------------------------

ax.text(
    5,
    9.55,
    "Principles of",
    fontsize=28,
    ha="center",
    va="top",
    fontweight="bold",
    color=MAROON,
    family="sans-serif",
)

ax.text(
    5,
    8.75,
    "Macroeconomics",
    fontsize=27,
    ha="center",
    va="top",
    fontweight="bold",
    color=MAROON,
    family="sans-serif",
)

# Change this to your actual course number
ax.text(
    5,
    0.55,
    "ECO 112",
    fontsize=24,
    ha="center",
    va="bottom",
    fontweight="normal",
    color=MAROON,
    family="sans-serif",
)

# ------------------------------------------------------------------
# Clean up and save
# ------------------------------------------------------------------

ax.axis("off")

plt.tight_layout()

plt.savefig(
    "eco112_icon_text.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white",
)

plt.show()
```
